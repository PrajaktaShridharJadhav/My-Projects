#### Problem : 17

17. A soft drink company claims that the average volume of juice in its bottles is 500 ml.
A quality inspector randomly selects 40 bottles and finds a sample mean of 495 ml with a standard deviation of 12 ml.

Task:

1. State the null and alternative hypotheses.

2. Identify the appropriate test.

3. Test the claim at a 5% significance level.

4. Write a conclusion in context.

 1. State the null and alternative hypotheses.

H0(null hypotheses) : 

                   the average volume of juice in its bottle is 500 ml.
                                        H0 = μ = 500

H1(alternate hypotheses) : 
                         
                   the average volume of juice in its  bottle is not equal to 500 ml.
                                         H1 = μ != 500

2. Identify the appropriate test.

The appropriate test for this case is One Sample T-test.

```python
# 3. Test the claim at a 5% significance level.

import numpy as np
from scipy import stats

sample_mean = 495
population_mean = 500
sample_std = 12
n = 40
significance = 0.05

t_stat = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df=n-1))

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < significance:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

4. Write a conclusion in context.

Interpretation of Results and conclusion

If p-value < 0.05 → Reject HO
If p-value !> 0.05 → Fail to reject H0

At the 5% significance level, there is sufficient statistical evidence to reject the company’s claim that 
the average volume of juice in the bottles is 500 ml.

The data suggests that the true mean volume is significantly different (and lower) than 500 ml.

#### Problem: 18

18. A city health department claims that 70% of households segregate waste properly. 
A random survey of 200 households shows that 128 households follow proper waste segregation.     

Task:

  1. Formulate H0 and H1.

  2. Choose the correct hypothesis test

  3. Test the claim at α = 0.05.

  4. Interpret the result in simple language.

1. Formulate H0 and H1.

     null hypotheses (H0)     = p = 70%
   
     alternate hypothesis(H1) = P != 70%

2. Choose the correct hypothesis test

The correct hypothesis test for this scenario is One-sample Z-test for proportion.

```python
# 3. Test the claim at α = 0.05.

p0 = 0.70
x = 128
n = 200
alpha = 0.05

p_hat = x / n

z = (p_hat - p0) / np.sqrt((p0 * (1 - p0)) / n)

p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print("Sample proportion:", p_hat)
print("Z-statistic:", z)
print("p-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

4. Interpret the result in simple language.

p-value = 0.064
p-value > 0.05
so we have failed to reject null hypothesis.

At the 5% significance level, there is not enough statistical evidence to reject the city health department’s 
claim that 70% of households segregate waste properly.

Although the sample proportion (64%) is lower than 70%, the difference is not statistically significant.

Task:

   1. Write the null and alternative hypotheses.

   2. Identify the appropriate hypothesis test.

   3. Test the claim at α = 0.05.

   4. Explain your conclusion.    

1. Write the null and alternative hypotheses.

    H0 = No average weight reduction

    H1 = Average weight is reduced

2. Identify the appropriate hypothesis test.

paired T-TEST will be used in this scenario as the sample is small and the test is performed at same individuals.

```python
# 3. Test the claim at α = 0.05.

import numpy as np
from scipy import stats


before = np.array([78, 85, 90, 72, 88, 95, 80, 76, 84, 92, 89, 86, 91, 83, 87]) #weights are taken randomly.
after  = np.array([76, 83, 88, 71, 86, 92, 78, 75, 82, 89, 87, 84, 89, 81, 85])


t_stat, p_value = stats.ttest_rel(before, after)

p_value = p_value / 2

print("t-statistic:", t_stat)
print("p-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

4. Explain your conclusion.

As the p-value < 0.05:


At the 5% significance level, there is sufficient evidence to conclude that the new workout program reduces body weight.

#### Problem: 20

20. A researcher wants to check whether gender and preference for online learning are independent.
A survey of 120 students records their gender (Male/Female) and preference (Online/Offline).

Task:

    1. State H0 and H1.

    2. Identify the test to be used.

    3. Conduct the test at 5% significance level.

    4. Interpret the result in real-world terms.

1. State H0 and H1.

      H0 = gender and preference for online learning are independent.

      H1 = gender and preference for online learning are not independent and have some association.

2. Identify the test to be used.

The perfect test for this scenario is chi-square test of independence as the data is categorical.

first we need to assume a contingency table .

eg: survey of 120 students.

              online         offline       total 
              
      male  : [ 40            20             60  ]
      
     female : [ 30            30             60  ]
     
     total  : [ 70            50            120  ]

```python
# 3. Conduct the test at 5% significance level.

import numpy as np
from scipy.stats import chi2_contingency

data = np.array([[40, 20],
                 [30, 30]])

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p_value)
print("Expected frequencies:\n", expected)

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
